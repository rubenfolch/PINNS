```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams.update({
    'font.size': 14,           # Tamaño base de la fuente
    'axes.labelsize': 16,      # Tamaño de las etiquetas de los ejes (xlabel, ylabel)
    'axes.titlesize': 18,      # Tamaño del título de la gráfica
    'xtick.labelsize': 14,     # Tamaño de los números de la escala X
    'ytick.labelsize': 14,     # Tamaño de los números de la escala Y
    'legend.fontsize': 14,     # Tamaño del texto de la leyenda
    'figure.titlesize': 20     # Tamaño del título general de la figura (si usas suptitle)
})
# Rango de entrada
x = np.linspace(-4, 4, 1000)

# Definición de funciones de activación
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)
elu = np.where(x > 0, x, (np.exp(x) - 1))


# Trazo de cada función con colores y estilos diferenciados
plt.plot(x, sigmoid, label="Sigmoide", color="#1f77b4", linewidth=2.2)
plt.plot(x, tanh, label="$\mathrm{tanh}$", color="#ff7f0e", linewidth=2.2)
plt.plot(
    x,
    relu,
    label="ReLU",
    color="#2ca02c",
    linewidth=2.2,
    linestyle="--",
)

plt.plot(
    x,
    elu,
    label="ELU",
    color="#9467bd",
    linestyle=":",
    linewidth=2,
)

# Ejes de referencia en el origen
plt.axhline(0, color="black", linewidth=0.8, linestyle=":")
plt.axvline(0, color="black", linewidth=0.8, linestyle=":")

# Formato de la gráfica
plt.xlabel("x")
plt.ylabel("f(x)")
plt.xlim(-4, 4)
plt.ylim(-2, 3)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend(loc="upper left", framealpha=0.9)

# Guardar figura para la memoria
plt.savefig("funciones_activacion.png", dpi=300, bbox_inches="tight")
```
